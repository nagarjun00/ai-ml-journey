```python
import numpy as np

# 1. Create a vector and a matrix
v = np.array([1, 2, 3])
m = np.array([[1, 2, 3], [4, 5, 6]])
print(v.shape, m.shape)

# 2. Matrix multiplication — the core operation behind every layer
w = np.array([[0.5, 0.2, 0.1], [0.3, 0.4, 0.6]])
result = np.dot(w, v)
print(result)

# 3. Write ReLU yourself
def relu(x):
    return np.maximum(0, x)
print(relu(np.array([-2, -1, 0, 1, 2])))

# 4. sigmoid 
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
print(sigmoid(np.array([-2, -1, 0, 1, 2])))

# 5. Simulating A neuron to check if it is  doing its actual job
inputs = np.array([0.5, 0.8, 0.2])
weights = np.array([0.1, 0.4, 0.3])
bias = 0.05
neuron_output = relu(np.dot(weights, inputs) + bias)
print("neuron output:", neuron_output)
```
